```python
import pandas as pd

# 1. Lê o arquivo REAL gerado pela Raspberry Pi
df = pd.read_csv('relatorio_final.csv')

# Listas vazias para guardar as datas e horas separadas
lista_datas = []
lista_horas = []

# Olha linha por linha da coluna horario e separa no espaço em branco
for horario_completo in df['horario']:
    # Transforma em texto (para evitar erro de valores nulos) e separa no espaço
    partes = str(horario_completo).split(' ')
   
    # Verificação de segurança: garante que conseguiu separar em 2 pedaços (data e hora)
    if len(partes) == 2:
        lista_datas.append(partes[0])  # Pega a data
        lista_horas.append(partes[1])  # Pega a hora
    else:
        # Se a linha estiver quebrada ou vazia, coloca um valor padrão para não travar
        lista_datas.append("Data_Erro")
        lista_horas.append("Hora_Erro")

# Guarda essas listas como colunas novas na tabela
df['data_limpa'] = lista_datas
df['hora_limpa'] = lista_horas

# Se alguma linha quebrou, a gente já limpa ela da tabela final
df = df[df['data_limpa'] != "Data_Erro"]

print("Dados reais carregados e organizados com sucesso!")
display(df.head())
```
